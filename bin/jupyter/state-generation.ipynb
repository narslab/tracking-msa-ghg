```python
import pandas as pd
import numpy as np
import os
import string
from scipy.stats import qmc
```

```python
# Number of samples you want to generate
num_samples = 10000  # You can adjust this
num_uncertainties = 9 #(10 dimensions, one for each parameter)

# Initialize the Latin Hypercube Sampling generator 
sampler = qmc.LatinHypercube(d=num_uncertainties, seed=42)  # 
lhs_samples = sampler.random(n=num_samples)  # Generate random samples
```

```python
bounds = pd.read_csv('../../data/tidy/uncertainty-ranges.csv')
bounds
```

```python
samples = qmc.scale(lhs_samples, bounds.LB, bounds.UB)
samples_df = pd.DataFrame(samples)
samples_df.columns = bounds.Parameter
samples_df
```

```python
samples_df.to_csv('../../results/lhs-samples.csv', index=False)
```
